#### Plot Axis(플롯 축)

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
print(plt.style.available)
plt.style.use('bmh')
plt.plot(np.random.randn(50).cumsum(),label='A')
plt.plot(np.random.randn(50).cumsum(),label='B')
plt.plot(np.random.randn(50).cumsum(),label='C')
plt.plot(np.random.randn(50).cumsum(),label='D')
plt.title('Random Graph')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
```

```python

plt.plot(np.random.randn(50));
plt.xlim(-1,100)
```

```python
plt.plot(np.random.randn(50));
plt.xlim(-1,100)
plt.ylim(-4,4)
```

```python
plt.plot(np.random.randn(50));
plt.axis([-1,100,-4,4])
```

```python
plt.plot(np.random.randn(50));
plt.axis([-1,100,-4,4])
plt.axis('tight')
```

```python
plt.plot(np.random.randn(50));
plt.axis([-1,100,-4,4])
plt.axis('equal')
```

```python
plt.plot(np.random.randn(50))
plt.title('Random Graph')
plt.xlabel('x')
plt.ylabel('y')
```

```python
plt.plot(np.random.randn(50).cumsum(),label='A')
plt.plot(np.random.randn(50).cumsum(),label='B')
plt.plot(np.random.randn(50).cumsum(),label='C')
plt.plot(np.random.randn(50).cumsum(),label='D')
plt.title('Random Graph')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
```
