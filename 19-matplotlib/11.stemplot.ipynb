#### Stem plot(줄기 - 잎 그래프)
- 데이터의 연속적인 값과 개별 측정값을 보존하면서 전체적인 분포 모양을 보여주는 그래프
- 특징
    - 연속 / 신호 데이터 시각화 : 이산신호처리나 시간 흐름에 따른 데이터 이산점 표시에 쓰임
    - 장점 : 개별 데이터 포인트의 위치와 크기를 명확히 파악가능

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0.1, 2*np.pi, 41)
y = np.exp(np.sin(x)) - 1

plt.stem(x, y, linefmt='crimson', markerfmt='rD', basefmt='r:')
```

#### Box Plot
- 데이터의 통계량(5수치 요약: 1사분위수, 중앙값, 3사분위수, 최댓값)을 상자 형태로 그림
- 구조
    - 상자(Box) : 데이터의 중앙 50%범위를 나타냄
    - 중앙값 선 : 상자 내부에 선으로 데이터의 중앙값 표시
    - 수염(Whisker) : 정상 데이터의 최댓값과 최솟값 범위까지 늘어난 선
    - 점 : 수염의 범위를 벗어난 극단값을 이상치 점으로 표시

- 장점 : 여러 그룹간의 데이터 분포, 편향성, 이상치 존재 여부를 비교할 때

```python
r = []
r.append(np.random.normal(loc=0, scale=0.5, size= 100))
r.append(np.random.normal(loc=0.5, scale=3.5, size= 100))
r.append(np.random.normal(loc=0, scale=2.5, size= 100))
r.append(np.random.normal(loc=-0.3, scale=0.5, size= 100))
r.append(np.random.normal(loc=2, scale=1.5, size= 100))

f, ax = plt.subplots(1,1)
ax.boxplot(r)
```
