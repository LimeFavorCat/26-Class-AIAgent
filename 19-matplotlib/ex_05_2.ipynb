1. 다음 데이터를 이용하여 학생의 5일간 공부 시간을 선 그래프로

* 요구사항

선 스타일은 점선으로 지정
마커는 사각형으로 지정

3. 두 학생의 시험 점수를 하나의 그래프에서 비교하기
exam = [1, 2, 3, 4, 5]

student_A = [70, 75, 80, 85, 90]
student_B = [65, 72, 78, 82, 88]

* 요구사항
두 학생의 점수를 하나의 그래프에 표시
A 학생은 파란색 실선, 원형 마커를 사용
B 학생은 초록색 점선, 삼각형 마커를 사용
범례에는 Student A, Student B를 표시
범례 위치는 오른쪽 아래로 지정
y축 범위는 60~100으로 지정
제목은 Exam Score Comparison으로 지정

```python
import matplotlib.pyplot as plt

exam = [1, 2, 3, 4, 5]

student_A = [70, 75, 80, 85, 90]
student_B = [65, 72, 78, 82, 88]

print(plt.style.available)
plt.style.use('seaborn-v0_8-bright')

# plt.plot(exam, student_A,'b-o', label = 'Student A')
# plt.plot(exam, student_B,'g--^', label = 'Student B')
plt.plot(exam, student_A, color = 'blue', linestyle = '-', marker = 'o', label='Student A')
plt.plot(exam, student_B, color = (0,1,0), linestyle= '--', marker= '^', label='Student B')
plt.legend(loc=4)
plt.ylim(60,100)
plt.title('Exam Score Comparison')
```

4.다음 데이터를 이용하여 월별 매출 변화를 그래프로
month = [1, 2, 3, 4, 5, 6]
sales = [120, 150, 130, 180, 200, 230]

* 요구사항
그래프의 제목과 축 이름을 한글로 작성
선 색상은 보라색으로 지정
선 스타일은 일점쇄선으로 지정
마커는 다이아몬드로 지정
x축 범위는 1~6, y축 범위는 100~250으로 지정
제목 글자 크기는 16, x축과 y축 글자 크기는 12로 지정
범례에는 매출액을 표시



```python
import matplotlib.pyplot as plt

month = [1, 2, 3, 4, 5, 6]
sales = [120, 150, 130, 180, 200, 230]

plt.rcParams['font.family'] = 'Malgun Gothic'
font1 = {'family':'Malgun Gothic', 'size':16, 'color' : 'g'}
font2 = {'family':'Malgun Gothic', 'size':12, 'color' : '#050502'}
plt.plot(month, sales, color = 'purple', linestyle = '-.', marker = 'D')
# plt.axis([1,6,100,250])
plt.xlim(1,6)
plt.ylim(100,250)
plt.xlabel('월', fontdict = font2)
plt.ylabel('판매액', fontdict = font2)
plt.title('월별 판매액 추이', fontdict = font1)
```
