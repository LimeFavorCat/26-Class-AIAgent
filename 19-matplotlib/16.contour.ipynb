```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
a  = np.arange(-1,1,0.1)
x,y = np.meshgrid(a,a)
z = np.sin(x * y)

cs = plt.contour(x,y,z,levels=a)
```

```python
a  = np.arange(-1,1,0.1)
x,y = np.meshgrid(a,a)
z = np.sin(x * y)

cs = plt.contour(x,y,z,levels=a)
plt.clabel(cs, inline=2)
plt.colorbar();
```

```python
a  = np.arange(-1,1,0.001)
x,y = np.meshgrid(a,a)
z = np.sqrt(x**2 + y**2)
mv = np.linspace(z.reshape(-1,1).min(), z.reshape(-1,1).max(), 40)

cs = plt.contour(x,y,z,levels=mv)
plt.clabel(cs, inline=2)
plt.colorbar();
```

```python
cs = plt.contourf(x,y,z,levels=mv)
plt.clabel(cs, inline=2)
plt.colorbar();
```
