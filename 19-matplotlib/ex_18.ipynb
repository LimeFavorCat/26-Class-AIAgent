문제 1. 한 가게의 상품별 판매량 데이터를 사용해 막대그래프와 원그래프를 각각 그리기
products = ['Coffee', 'Tea', 'Juice', 'Water']
sales = [120, 80, 60, 40]

* 요구사항
1. 그래프 2개를 나란히 배치
2. 첫 번째 그래프는 bar plot으로 상품별 판매량을 표시.
3. 막대 색상은 상품마다 다르게 지정하고, 제목과 축 이름을 추가
4. 두 번째 그래프는 pie chart로 판매 비율을 표시
5. 원그래프에 퍼센트(autopct)와 상품 이름을 표시

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
products = ['Coffee', 'Tea', 'Juice', 'Water']
sales = [120, 80, 60, 40]

ax = plt.subplot(121)
ax.bar(products, sales, color =['blue','orange','green','red'])
ax.set_title('sales for product')
ax.set_xlabel('product')
ax.set_ylabel('sales')

ax = plt.subplot(122)
ax.pie(sales, startangle=90, labels=products, autopct= '%.1f%%')
```

문제 2. 다음 측정값으로 이산적인 값의 변화를 나타내는 stem plot과 연속적인 분포를 살펴보는 histogram을 그리기
x = np.arange(1, 11)
measurements = np.array([4, 7, 5, 8, 6, 9, 5, 7, 10, 6])

* 요구사항
1. 두 그래프를 나란히 배치
2. 왼쪽에는 stem plot을 그리고 제목과 축 이름을 추가
3. 오른쪽에는 histogram을 그리고 구간 수를 5개로 지정
4. 히스토그램 막대에 테두리를 검정색으로 표시

```python
x = np.arange(1, 11)
measurements = np.array([4, 7, 5, 8, 6, 9, 5, 7, 10, 6])

ax = plt.subplot(121)
ax.set_title('stem plot')
ax.set_xlabel('x')
ax.set_ylabel('measurements')
ax.stem(x,measurements)

ax = plt.subplot(122)
ax.hist(measurements, bins=5, edgecolor='k')
```

문제 3. 두 반의 시험 점수와 공부 시간·시험 점수 사이의 관계를 시각화
class_A = [65, 72, 78, 80, 85, 90, 92]
class_B = [55, 68, 70, 74, 76, 81, 95]

study_hours = [1, 2, 2.5, 3, 3.5, 4, 5, 6]
scores = [50, 55, 60, 65, 70, 78, 85, 92]

* 요구사항
1. 왼쪽 그래프에서 두 반의 점수를 box plot으로 비교
2. 오른쪽 그래프에서 공부 시간과 점수의 관계를 scatter plot으로 표시
3. 산점도 마커 크기를 70, 투명도를 0.7로 지정
4. 두 그래프에 제목과 축 이름을 추가

```python
class_A = [65, 72, 78, 80, 85, 90, 92]
class_B = [55, 68, 70, 74, 76, 81, 95]
classes= [class_A,class_B]
study_hours = [1, 2, 2.5, 3, 3.5, 4, 5, 6]
scores = [50, 55, 60, 65, 70, 78, 85, 92]
fig, ax = plt.subplots(1,2, sharey=True)

ax = plt.subplot(121)
ax.boxplot(classes)
ax.set_title('box plot')
ax.set_xlabel('class')
ax.set_ylabel('scores')


ax = plt.subplot(122)
ax.scatter(study_hours,scores, s=70, alpha=0.7)
ax.set_title('scatter plot')
ax.set_xlabel('study_hours')
```

문제 4. 두 신호가 공통 주파수 성분을 공유하는지 coherence plot으로 확인
* 조건
- 샘플링 주파수 fs = 500 Hz
- 시간 길이는 4초
- 두 신호 모두 40 Hz 사인파로 만들기

* 요구사항
1. 시간 배열 t를 만들기
2. 두 신호 x, y를 생성
3. coherence로 그리기.
4. Fs=fs, NFFT=256을 지정하고 제목과 축 이름 추가

```python
# 신호 데이터 생성
dt = 0.002
t = np.arange(0, 4, dt)
n1 = np.random.randn(len(t))
n2 = np.random.randn(len(t))

s1 = np.sin(np.pi * 40 * t) + n1
s2 = np.sin(np.pi * 40 * t) + n2

plt.cohere(s1, s2**2, NFFT=256, Fs = 1.0/dt)
plt.title('Coherence Plot')
```

문제 5. Pcolor로 2차원 값의 분포 표현하기
온도와 비슷한 가상의 값을 2차원 색상 격자로 표현

* 요구사항
1. x, y를 각각 -3부터 3까지 0.25 간격으로 만들기
2. np.meshgrid(x, y)로 좌표 격자 X, Y를 만들기
3. Z = np.sin(X) * np.cos(Y)를 계산
4. plt.pcolor()로 색상 격자를 그리고 colorbar를 추가
5. 제목과 x, y축 이름을 설정

```python
x = np.arange(-3,3,0.25)
y = np.arange(-3,3,0.25)

X, Y = np.meshgrid(x,y)
Z = np.sin(X) * np.cos(Y)
plt.pcolor(x,y,Z)
plt.colorbar()
plt.title('Temperature')
plt.xlabel('x')
plt.ylabel('y')
```

문제 6. Meshgrid와 Contour로 등고선 그리기
Z = X**2 + Y**2

* 요구사항
1. x와 y를 -3부터 3까지 100개의 점으로 생성
2. np.meshgrid()로 2차원 격자를 생성하기
3. plt.contour()로 등고선을 10개 수준(levels)으로 그리기
4. plt.clabel()로 등고선에 숫자 라벨을 표시
5. 축 비율을 동일하게 설정하고 제목과 축 이름을 추가

```python
x = np.linspace(-3,3,100)
y = np.linspace(-3,3,100)

X,Y = np.meshgrid(x,y)
z = np.sqrt(X**2 + Y**2)
print(z)
mv = np.linspace(z.reshape(-1,1).min(), z.reshape(-1,1).max(), 10)

cs = plt.contour(x,y,z,levels=mv)
plt.clabel(cs, inline=2)
plt.title('ground height')
```

문제 7. Meshgrid를 이용한 3D Surface와 등고선 투영
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)

요구사항
1. x, y를 -5부터 5까지 100개씩 생성하고 np.meshgrid()를 사용
2. fig.add_subplot(111, projection='3d')로 3D 축을 만들기
3. plot_surface()로 3D 표면을 그리기
4. contour()를 이용해 z=-1.5 평면에 등고선을 투영
5. 제목, x·y·z축 이름을 설정하고 colorbar를 추가

```python
x = np.linspace(-5,5,100)
y = np.linspace(-5,5,100)
X, Y = np.meshgrid(x,y)
R = X**2 + Y**2
Z = np.sqrt(R)

fig.clear()
ax1 = fig.add_subplot(121, projection='3d')
ax1.plot_surface(x,y,Z, cmap='OrRd')


ax1.figure

ax2 = fig.add_subplot(122)
ax2.contour(x,y,R,30)

ax2.figure
```
