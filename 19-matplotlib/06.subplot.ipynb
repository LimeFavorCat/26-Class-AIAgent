#### Multiple Subplot(다중플롯)

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.style.available
```

```python
plt.style.use('seaborn-v0_8-whitegrid')
plt.plot([1,2,3,4,5,6])
```

```python
ax1 = plt.axes()
ax2 = plt.axes([0.65, 0.5, 0.2, 0.3])
```

```python
for i in range(1,10):
    plt.subplot(3, 3, i)
```

```python
for i in range(1,4):
    plt.subplot(3, 3, i*2)
```

```python
for i in range(1,10):
    plt.subplot(3, 3, i)
    plt.text(0.5,0.5,str((3,3,i)), ha='center')
```

```python
fig, ax = plt.subplots(3,3,sharex='col', sharey='row')
```

```python
fig
```

```python
ax
```

```python
ax[0,2].plot([1,2,3,2,1,1],'r-o')
ax[1,1].plot([5,4,4,3,4,2,1],'g:x')
ax[2,1].plot([6,4,2,3,5],'g:x')
fig
```

```python
grid = plt.GridSpec(2,3, wspace=0.4, hspace=0.4)
plt.subplot(grid[0,0])
plt.subplot(grid[0,1:])
plt.subplot(grid[1,:2])
plt.subplot(grid[1,2])
```

```python
plt.figure(figsize=(5,6))

x= range(1,21)
columns = [np.random.randn(20) * i for i in range(1,7)]
columns
```

```python
i = 0
for c in columns:
    i += 1

    plt.subplot(3,2,i)
    plt.plot(x, c, '-o')
    plt.axis([-0.5,20.5,c.min()-0.5,c.max()+0.5])
```
