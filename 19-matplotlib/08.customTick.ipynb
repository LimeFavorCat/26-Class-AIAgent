#### 눈금 맞춤 (Customizing Ticks)
- 위치 & 포맷 지시자

|위치 & 포맷 지시자|설명
|:---:|:---:
| NullLocator | 눈금표시 안함 |
| FixedLocator | 명시적으로 지정한 좌표 리스트에만 눈금 배치 |
| IndexLocator | 인데스 플롯을 위한 위치 지시자 |
| LinearLocator | min에서 max까지 일정 간격 눈금 |
| LogLocator | min에서 max까지 로그 간격으로 눈금 배치 |
| MultipleLocator | 지정한 고정 간격(base)마다 눈금 배치 |
| MaxNLocator | 최대 n개(nbins) 이하의 눈금이 보기 좋게 들어가도록 자동 배치 |
| AutoLocator | 그래프 범위에 맞추어 기본(Default) 눈금 자동 배치 |
| AutoMinorLocator | 보조 눈금에 대한 위치 지시자 |
| NullFormatter | 눈금선은 남겨두고 라벨 텍스트만 제거 |
| PercentFormatter | 수치를 백분율(%) 형식으로 변환하여 표시 |
| IndexFormatter | 레이블 목록에서 문자열 설정 |
| FixedFormatter | 레이블에 직접 문자열 설정 |
| FuncFormatter | 사용자 정의 함수로 레이블 설정 |
| FormatStrFormatter | printf 포맷 문자열(%...) 적용 |
| ScalarFormatter | 스칼라값에 대한 포맷 지시자 |
| LogFormatter | 로그 축에 대한 기본 포맷 지시자 |

```python
import matplotlib.pyplot as plt
ax = plt.axes()
ax.yaxis.set_major_locator()
```

```python
fig, ax = plt.subplots(3, 3, sharex=True, sharey=True)
```

```python
for axi in ax.flat:
    axi.xaxis.set_major_locator(plt.MaxNLocator(4))
    axi.yaxis.set_major_locator(plt.MaxNLocator(10))
fig
```

```python
import numpy as np
x= np.linspace(-np.pi,np.pi,1000)
y = np.sin(x)
plt.plot(x, y)

ax = plt.gca()
ax.spines['bottom'].set_position(('data', 0))
ax.spines['top'].set_position(('data', 0))
ax.spines['left'].set_position(('data', 0))
ax.spines['right'].set_position(('data', 0))

ax.xaxis.set_ticks_position('bottom')
ax.yaxis.set_ticks_position('left')
```

```python
x = np.linspace(1, 10)
y = [10**el for el in x]
print(x)
print(y)

fig = plt.figure(figsize=(10,8))
ax1= fig.add_subplot(2,2,1)
ax1.plot(x,y, 'y')
ax1.set_yscale('log')
ax1.set_title(r'Logarithmic plot of $ {10}^{x} $')
ax1.set_ylabel(r'$ {y} = {10}^{x} $')
plt.grid(which='both', axis='both')

ax2 = fig.add_subplot(2,2,2)
ax2.plot(x,y,'--r')
ax2.set_yscale('linear')
ax2.set_title(r'Linear plot of $ {10}^{x} $')
ax2.set_ylabel(r'$ {y} = {10}^{x} $')
plt.grid(which='both', axis='both')


ax3 = fig.add_subplot(2,2,3)
ax3.plot(x,y,'--r')
ax3.set_yscale('linear')
ax3.set_title(r'Linear plot of $ {2}^{x} $')
ax3.set_ylabel(r'$ {y} = {2}^{x} $')
plt.grid(which='both', axis='both')


ax4 = fig.add_subplot(2,2,4)
ax4.plot(x,y,'--r')
ax4.set_yscale('linear')
ax4.set_title(r'Linear plot of $ {2}^{x} $')
ax4.set_ylabel(r'$ {y} = {2}^{x} $')
plt.grid(which='both', axis='both')
```
