```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = [10,30,50,20,60]
plt.rc('font',family = 'Malgun Gothic')
categories = ['블루베리','오렌지','샤인머스켓','사과','포도']
plt.pie(data, labels=categories, autopct='%.1f')
```

```python
data = [10,30,50,20,60]
plt.rc('font',family = 'Malgun Gothic')
categories = ['블루베리','오렌지','샤인머스켓','사과','포도']
plo = [0.05 for _ in range(5)]
plo[4]= 0.2
plt.pie(data, explode=plo labels=categories, autopct='%.1f%%')
```

```python
data = [10,30,50,20,60]
plt.rc('font',family = 'Malgun Gothic')
categories = ['블루베리','오렌지','샤인머스켓','사과','포도']
plo = [0.05 for _ in range(5)]
plo[4]= 0.2
plt.pie(data, explode=plo, startangle=90, labels=categories, autopct='%.1f%%')
```

```python
plt.figure(figsize=(6,6))

data = [10,30,50,20,60]
plt.rc('font',family = 'Malgun Gothic')
categories = ['블루베리','오렌지','샤인머스켓','사과','포도']
plo = [0 for _ in range(5)]
plo[4]= 0.1
plt.pie(data, explode=plo, startangle=90, labels=categories, autopct='%.1f%%')

# 중앙에 흰색 원 추가
center_circle = plt.Circle((0,0),0.3, fc='white')
fig = plt.gcf()
fig.gca().add_artist(center_circle)
plt.title('2D 도넛 차트')
```
