```python
import matplotlib.pyplot as plt
import numpy as np
```

```python

plt.style.use('seaborn-v0_8-dark')
plt.rcParams.update({
    'axes.facecolor' : 'white',
    'grid.color' : 'gray',
    'grid.alpha':0.5
})

data = np.random.normal(0,50,50000)
plt.hist(data, bins=30,color = 'purple', edgecolor = '#DB9CFF', linewidth = 2);
```

```python
x1 = np.random.normal(0,1,10000)
x2 = np.random.normal(-5,3,10000)
x3 = np.random.normal(5,2,10000)


d= dict(histtype='stepfilled', alpha=0.3, bins=50)

plt.hist(x1, **d);
plt.hist(x2, **d);
plt.hist(x3, **d);

```
