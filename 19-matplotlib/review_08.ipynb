```python
import matplotlib as mpl
import matplotlib.pyplot as plt # <- 주로 이것만 사용함
import numpy as np
```

```python
plt.style.available
```

```python
plt.style.use('seaborn-v0_8-whitegrid')
plt.plot(np.random.randn(50).cumsum())
plt.plot(np.random.randn(50).cumsum())
plt.title('Random plot')
```

```python
plt.rc('font', family = 'Malgun Gothic')
#plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] =False
plt.title('한글 폰트 사용하기')
plt.annotate('한글 주석', xy=(0.4,0.6))
plt.xlabel('한글 이름', size = 7)
```

#### plot 그리기

```python
plt.plot([1,2,3,4,5]) # 배열 하나만 넣을 경우 y값으로 취급 (x = 0,1,2,3...)
plt.plot([2,3,4,5,6],[1,2,3,4,5]) # 배열 두개를 넣을 경우 x,y값으로 취급
x = np.linspace(0,6,100)
plt.plot(x,np.sin(x)) # np와 같이 사용하면 다양한 그래프 가능
```

```python
# plt.plot() 함수에서 자주 사용하는 매개변수
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] =False
plt.plot([1,2,3,4,5], label = '대각선', color = 'black', linestyle = '-', marker  = 'o' ,linewidth = 3, markeredgewidth = 3)
plt.plot([2,3,4,5,6],[1,2,3,4,5], 'b:D', label = '대각선2') # 선 양식을 간단하게 표현가능
x = np.linspace(0,6,100)
plt.plot(x,np.sin(x), '--', color = 'crimson', label = '삼각함수'); # 세미콜론을 붙여서 출력하면 메세지 없이 그래프만 출력
plt.legend() # label 출력하기
```

#### 축 범위 설정하기

```python
plt.xlim(3,5) # plt.xlim(xmin, xmax)
plt.ylim(1,1.2)
```

```python
plt.axis([1,3,100,200]) # plt.axis([xmin,xmax, ymin,ymax])
```

#### 제목 및 X,Y 라벨 표시하기

```python
plt.title('타이틀에 표시할 텍스트')
plt.xlabel('X축에 표시할 텍스트')
plt.ylabel('Y축에 표시할 텍스트')
```

```python
plt.plot([1,2,3,4,5], label = '대각선')
plt.plot([2,3,4,5,6],[1,2,3,4,5], label = '대각선2')
x = np.linspace(0,6,100)
plt.plot(x,np.sin(x), label = '삼각함수')
plt.legend(); # label 출력하기
```

#### 그래프 여러개 그리기

```python
# 좌표를 지정해 그리기
ax1 = plt.axes()
ax2 = plt.axes([0.65, 0.5, 0.2, 0.3])
ax3 = plt.axes([0.0, 0.1, 0.8, 0.3])
ax4 = plt.axes([0.0, 0.1, 0.4, 0.3]) # x 좌표, y 좌표, x 길이, y 길이  (figure.transFigure 좌표계)
ax4.plot([1,3,5])
```

```python
for i in range(1,10):
    plt.subplot(3, 3, i) # 화면을 3,3 격좌로 나누었을 떄 i번째칸에 그리기


for i in range(1,10):
    temp = plt.subplot(4, 4, i)
    temp.plot([1,2,3])
```
